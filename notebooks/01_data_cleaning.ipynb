```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_csv("D:/Retail Supply Chain Intelligence Platform/data/raw/olist_customers_dataset.csv")

orders = pd.read_csv("D:/Retail Supply Chain Intelligence Platform/data/raw/olist_orders_dataset.csv")

order_items = pd.read_csv("D:/Retail Supply Chain Intelligence Platform/data/raw/olist_order_items_dataset.csv")

payments = pd.read_csv("D:/Retail Supply Chain Intelligence Platform/data/raw/olist_order_payments_dataset.csv")

reviews = pd.read_csv("D:/Retail Supply Chain Intelligence Platform/data/raw/olist_order_reviews_dataset.csv")
```

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in datasets.items():
    print("="*60)
    print(name)
    print(df.info())
```

```python
for name, df in datasets.items():
    print("\n", name)
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print(name)
    print(df.duplicated().sum())
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
```

```python
customers.to_csv(
    "D:/Retail Supply Chain Intelligence Platform/data/processed/customers_clean.csv",
    index=False
)

orders.to_csv(
    "D:/Retail Supply Chain Intelligence Platform/data/processed/orders_clean.csv",
    index=False
)

order_items.to_csv(
    "D:/Retail Supply Chain Intelligence Platform/data/processed/order_items_clean.csv",
    index=False
)

payments.to_csv(
    "D:/Retail Supply Chain Intelligence Platform/data/processed/payments_clean.csv",
    index=False
)

reviews.to_csv(
    "D:/Retail Supply Chain Intelligence Platform/data/processed/reviews_clean.csv",
    index=False
)
```
